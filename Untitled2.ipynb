```python
peas = {
    "Chess Playing Program": {
        "Performance": "Win the game",
        "Environment": "Chessboard and opponent",
        "Actuators": "Move chess pieces",
        "Sensors": "Chessboard position and clock"
    },
    "Online English Tutor": {
        "Performance": "Improve student English",
        "Environment": "Online classroom",
        "Actuators": "Display lessons and questions",
        "Sensors": "Student answers and voice input"
    },
    "Parcel Delivery Drone": {
        "Performance": "Deliver parcels safely and quickly",
        "Environment": "Roads, buildings, weather, and airspace",
        "Actuators": "Propellers and package release mechanism",
        "Sensors": "GPS, camera, and obstacle sensors"
    }
}

for agent, details in peas.items():
    print("\n", agent)
    for key, value in details.items():
        print(key + ":", value)
```

```python
import random

ACTIONS = ["Left", "Right", "Suck"]


def run_agent(agent_type, initial_state):
    state = initial_state.copy()
    position = 0
    score = 0

    for _ in range(20):
        if agent_type == "Random":
            action = random.choice(ACTIONS)
        else:
            if state[position] == 1:
                action = "Suck"
            else:
                action = "Right" if position == 0 else "Left"

        if action == "Suck":
            if state[position] == 1:
                state[position] = 0
                score += 10
            else:
                score -= 1

        elif action == "Left":
            score -= 1
            position = 0

        elif action == "Right":
            score -= 1
            position = 1

    return score


random_scores = []
reflex_scores = []

for i in range(10):
    initial = [random.randint(0, 1), random.randint(0, 1)]

    random_scores.append(run_agent("Random", initial))
    reflex_scores.append(run_agent("Reflex", initial))

print("Random Agent Average:",
      sum(random_scores) / 10)

print("Simple Reflex Agent Average:",
      sum(reflex_scores) / 10)
```

```python
import random


def simple_reflex(initial):
    state = initial.copy()
    position = 0
    score = 0

    for _ in range(20):
        if state[position] == 1:
            state[position] = 0
            score += 10
        else:
            score -= 1
            position = 1 - position

    return score


def model_based(initial):
    state = initial.copy()
    position = 0
    known_clean = set()
    score = 0

    for _ in range(20):
        if state[position] == 1:
            state[position] = 0
            known_clean.add(position)
            score += 10

        else:
            known_clean.add(position)

            if len(known_clean) == 2:
                # NoOp: stop when both locations are clean
                break

            score -= 1
            position = 1 - position

    return score


reflex_total = 0
model_total = 0

for _ in range(10):
    initial = [random.randint(0, 1),
               random.randint(0, 1)]

    reflex_total += simple_reflex(initial)
    model_total += model_based(initial)

print("Simple Reflex Average:",
      reflex_total / 10)

print("Model-Based Average:",
      model_total / 10)
```
